```python
X=data.iloc[:,:-1]
```

# One-hot encoding

```python
# Sample dataframe
#data = {'dependent': ['lactate','notlactate']}

# Apply one-hot encoding
encoded_df = pd.get_dummies(data, columns=['I79'], dtype=int)

print(encoded_df)
```

# Model for not lactation

```python
model1=LinearRegression()
model1.fit(X, encoded_df['I79_0'])
```

# Model for lactation

```python
model2=LinearRegression()
model2.fit(X, encoded_df['I79_1'])
```
